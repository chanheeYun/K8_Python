```python
N = int(input())
for i in range(1, 10):
  print(f'{N} * {i} = {N*i}')
```

```python
T = int(input())
for i in range(T):
  A, B = map(int, input().split())
  print(A + B)
```

```python
num = int(input())
total = sum([i for i in range(1, num+1)])
print(total)
```

```python
X = int(input())
N = int(input())
total = 0
for i in range(N):
  a, b = map(int, input().split())
  total += a*b
print('Yes' if X == total else 'No')
```
